```python
from sklearn.linear_model import LinearRegression
x=[[1],[2],[3],[4],[5]]
y=[[3],[6],[9],[12],[15]]
model=LinearRegression()
model.fit(x,y)
print(model.predict([[12]]))
```

```python
m=model.coef_[0]
c=model.intercept_
print('slope:',m)
print('slope:',c)
print('prediction:',model.predict([[12]]))
```
